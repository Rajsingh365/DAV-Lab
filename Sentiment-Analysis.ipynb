```python
from textblob import TextBlob
```

```python
text = input("Enter a sentence: ")
```

```python
text='This product isn\'t worth it'
```

```python
blob = TextBlob(text)
sentiment = blob.sentiment.polarity

if sentiment>0:
  print('Positive')
elif sentiment<0:
  print('Negative')
else: 
  print('Neutral')
```
