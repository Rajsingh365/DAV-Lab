```python
import numpy as np
import pandas as pd

data={
  'X': [1,2,3,4,5],
  'y': [3,4,7,9,21]
}
df = pd.DataFrame(data=data)
```

```python

X=df['X'].values
y=df['y'].values

X_mean=np.mean(X)
y_mean=np.mean(y)

num=sum((X-X_mean) * (y-y_mean))
den=sum((X-X_mean) ** 2)

m = num/den
c = y_mean - m*X_mean

def predict(x):
  return m*x + c
```

```python
import matplotlib.pyplot as plt
y_pred = m*X + c
plt.scatter(X, y=y)
plt.plot(X, y_pred, color = 'red')
plt.show()
```

```python
# Test prediction
print(predict(2.5))
```

```python
# rmse
def rmse(y, y_pred):
    n = len(y)
    squared_errors = (y - y_pred) ** 2
    mse = np.sum(squared_errors) / n
    rmse = np.sqrt(mse)
    return rmse

def r2_score(y, y_pred):
    sse = np.sum((y - y_pred) ** 2)
    sst = np.sum((y - np.mean(y)) ** 2)
    return 1 - (sse / sst)

# Testing the model
# test_x = int(input("Enter feature value: "))
test_x = 7
predicted_y = predict(test_x)
print(f"Predicted Y for X = {test_x} is {predicted_y}")
print(f"RMSE is {rmse(y, y_pred)}")
print("R² Score:", r2_score(y, y_pred))
```
